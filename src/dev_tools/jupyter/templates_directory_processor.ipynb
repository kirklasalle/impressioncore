# Templates Directory Processor

This notebook automates tasks related to the templates directory:
1. Creates a directory listing
2. Saves the listing to a Markdown file
3. Updates base.html with the listing information
4. Updates server.py to reflect the changes

This automation helps maintain consistency between the template files and their references in the application.

## Import Required Libraries

We'll need libraries for file system operations, path handling, and content manipulation.

```python
# Import Required Libraries
import os
import pathlib
from pathlib import Path
import datetime
import re
import shutil
```

## List Files in Directory

We'll scan the templates directory and create a structured listing of all files.

```python
# Set templates directory path
templates_dir = Path('src/web/templates')

# Check if directory exists
if not templates_dir.exists():
    print(f"Warning: Directory {templates_dir} does not exist. Creating it...")
    templates_dir.mkdir(parents=True, exist_ok=True)
    print(f"Created directory: {templates_dir}")

# Get all files in the templates directory
template_files = []
for file_path in templates_dir.glob('**/*.*'):
    if file_path.is_file():
        relative_path = file_path.relative_to(templates_dir)
        template_files.append({
            'name': file_path.name,
            'path': str(relative_path),
            'extension': file_path.suffix,
            'modified': datetime.datetime.fromtimestamp(file_path.stat().st_mtime).strftime('%Y-%m-%d %H:%M:%S')
        })

# Display found templates
print(f"Found {len(template_files)} template files:")
for file in template_files:
    print(f"- {file['path']}")
```

## Write Directory Listing to Markdown

Generate a well-formatted Markdown file with the directory listing.

```python
# Define the output markdown file path
markdown_file = Path('src/web/docs/templates_listing.md')

# Ensure the directory exists
markdown_file.parent.mkdir(parents=True, exist_ok=True)

# Generate markdown content
markdown_content = f"""# Templates Directory Listing

> Generated on: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}

This document contains a listing of all template files in the `src/web/templates` directory.

## Files

| Filename | Path | Extension | Last Modified |
|----------|------|-----------|---------------|
"""

# Add each file to the markdown
for file in template_files:
    markdown_content += f"| {file['name']} | {file['path']} | {file['extension']} | {file['modified']} |\n"

# Add section for template usage
markdown_content += """
## Template Usage

The templates in this directory are used by the Flask application to render HTML pages.
Each template extends from `base.html` and provides specific content for different routes.

### Common Templates

- `base.html`: The main template that defines the overall HTML structure
- `index.html`: The home page template
- `error.html`: Template for error pages

"""

# Write to file
with open(markdown_file, 'w') as f:
    f.write(markdown_content)

print(f"Successfully wrote listing to {markdown_file}")
```
